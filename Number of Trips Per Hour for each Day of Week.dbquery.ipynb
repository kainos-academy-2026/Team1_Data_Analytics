```sql
-- Finding the busiest hour of each day of the week
SELECT COUNT(*) as num_trips, 
       hour(f.pickup_due_timestamp) AS pickup_hour, 
       d.day_name
FROM students_data.team1_taxi.fact_trip f
INNER JOIN students_data.team1_taxi.dim_date d
ON f.trip_start_date_key = d.date_key
GROUP BY hour(f.pickup_due_timestamp), d.day_name
ORDER BY num_trips DESC;
```
