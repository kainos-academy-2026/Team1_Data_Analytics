```sql
-- Change inline with new gold schema
SELECT d.day_name,
       ROUND(AVG(f.wait_time_minutes), 2) AS avg_wait_min,
       ROUND(AVG(f.trip_duration_minutes), 2) AS avg_trip_duration
FROM students_data.team1_taxi.fact_trip f
JOIN students_data.team1_taxi.dim_date d ON f.date_key = d.date_key
WHERE f.wait_time_minutes IS NOT NULL
GROUP BY d.day_name
```
