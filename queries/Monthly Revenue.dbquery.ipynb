```sql
SELECT d.year, d.month_name, d.month,
       SUM(f.fare_amount) AS total_revenue,
       COUNT(f.trip_key) AS total_trips
FROM students_data.team1_taxi.fact_trip f
JOIN students_data.team1_taxi.dim_date d ON f.date_key = d.date_key
GROUP BY d.year, d.month_name, d.month
ORDER BY d.year, d.month
```
