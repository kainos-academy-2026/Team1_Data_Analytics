```python
from delta.tables import DeltaTable
from pyspark.sql.functions import *
from pyspark.sql.functions import col
```

```python
weather_raw_df = spark.read.csv(
    "/Volumes/students_data/diarmuid-gallagher/technologist-volume/weather_data_24hr.csv",
    header=True,
    inferSchema=True,
    mode="DROPMALFORMED"
)

# Replace spaces
for column in weather_raw_df.columns:
    weather_raw_df = weather_raw_df.withColumnRenamed(column, column.replace(" ", "_"))

weather_raw_df = weather_raw_df.withColumn("ingested_at", current_timestamp())
weather_raw_df = weather_raw_df.withColumn("source_file", col("_metadata.file_path"))


weather_raw_df.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("weather_raw")
```
